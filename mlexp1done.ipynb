```python
# Task 1: Import required libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Task 2: Load train.csv into a Pandas DataFrame
df = pd.read_csv("Titanic-Dataset (1) (1).csv")

# Display the first few rows (optional)
print(df.head())
```

```python
df.info()
```

### Task 7: Generate summary statistics

```python
summary_stats = df.describe()
display(summary_stats)

print(f"\nAge - Mean: {summary_stats.loc['mean', 'Age']:.2f}, Min: {summary_stats.loc['min', 'Age']:.2f}, Max: {summary_stats.loc['max', 'Age']:.2f}")
print(f"Fare - Mean: {summary_stats.loc['mean', 'Fare']:.2f}, Min: {summary_stats.loc['min', 'Fare']:.2f}, Max: {summary_stats.loc['max', 'Fare']:.2f}")
```

### Task 8: Identify missing values

```python
display(df.isnull().sum())
```

### Task 9: Fill missing Age values with the median

```python
median_age = df['Age'].median()
df['Age'].fillna(median_age, inplace=True)

print(f"Missing 'Age' values after filling with median ({median_age}):")
display(df['Age'].isnull().sum())
```

### Task 10: Filter surviving passengers (Survived == 1)

```python
survived_passengers_df = df[df['Survived'] == 1]

print("First 5 rows of surviving passengers DataFrame:")
display(survived_passengers_df.head())
```

### Task 11: Convert Fare column to a NumPy array

```python
fare_array = df['Fare'].to_numpy()

print("Fare column converted to NumPy array (first 10 elements):")
print(fare_array[:10])
print(f"Type of fare_array: {type(fare_array)}")
```

### Task 12: Calculate maximum and average Fare using np.max() and np.mean()

### Task 13: Plot histogram of Age

```python
plt.figure(figsize=(10, 6))
sns.histplot(df['Age'], bins=20, kde=True)
plt.title('Distribution of Age')
plt.xlabel('Age')
plt.ylabel('Count')
plt.grid(axis='y', alpha=0.75)
plt.show()
```

### Task 14: Plot bar chart of Survival Count

```python
plt.figure(figsize=(8, 5))
sns.countplot(x='Survived', data=df, palette='viridis')
plt.title('Survival Count (0 = Not Survived, 1 = Survived)')
plt.xlabel('Survived')
plt.ylabel('Count')
plt.xticks([0, 1], ['Not Survived', 'Survived'])
plt.show()
```

### Task 15: Plot scatter graph of Age vs Fare

```python
plt.figure(figsize=(12, 7))
sns.scatterplot(x='Age', y='Fare', hue='Survived', data=df, alpha=0.7)
plt.title('Age vs Fare with Survival Status')
plt.xlabel('Age')
plt.ylabel('Fare')
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```

```python
max_fare = np.max(fare_array)
avg_fare = np.mean(fare_array)

print(f"Maximum Fare: {max_fare:.2f}")
print(f"Average Fare: {avg_fare:.2f}")
```
